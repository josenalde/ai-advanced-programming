```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.cluster import KMeans
from sklearn.datasets import load_digits
```

```python
bunch = load_digits()
print(type(bunch))
```

```python
df = bunch.data
print(type(df))
```

```python
df.shape
```

```python
np.random.seed(10)
```

```python
n_clusters = 10 #sei de antemão que são digitos de 0 a 9
```

```python
model = KMeans(n_clusters=n_clusters, init='random')
```

```python
model.fit(df)
```

```python
classes_predicted = model.predict(df) #yhat
```

```python
classes_predicted
```

```python
for i in range(0, n_clusters):
  row = np.where(classes_predicted==i)[0]
  num = row.shape[0] #quantos elementos existem no grupo i
  r = np.floor(num/10.) #numero de linhas na figura de cada cluster
  r = r.astype(int)
  print('cluster ' + str(i) + ' com ' + str(num) + ' elementos')
  plt.figure(figsize=(10,10))
  for k in range(0, num): #itera sobre cada linha, cada figura de digito
    plt.subplot(r+1,10,k+1)
    image = df[row[k], ].reshape(8,8)
    plt.imshow(image, cmap='gray')
    plt.axis('off')
  plt.show()

```
